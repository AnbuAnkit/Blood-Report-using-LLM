```python
from dotenv import load_dotenv
import os
import json
from langchain_google_genai import ChatGoogleGenerativeAI
# %pip install importnb
load_dotenv()
```

```python
with open('blood report.txt', encoding="utf-8",mode='r') as file:
    blood_report = file.read()
```

```python
model = ChatGoogleGenerativeAI(
    model="gemma-4-31b-it",
    temperature=0.1,
)
```

```python
System_prompt = """
You are a helpful assistant that summarizes blood reports. You will be given a blood report and you will summarize it in a concise manner. You will also provide recommendations based on the blood report.
"""

Dietetary_prompt = """
Based on the blood report, suggest some dietary suggestions that can help improve the blood report results. Provide a list of foods to include and avoid in the diet.
"""
```

```python
response = model.invoke(System_prompt + "\n\n" + Dietetary_prompt + "\n\n" + blood_report)
```

```python
print(response.text)
```

```python
with open('blood_report_llm_output.json', encoding="utf-8",mode='w') as file:
    json.dump(response.content, file, indent=4)
```
